```python
import nltk
import pandas as pd
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer

# Download necessary NLTK datasets
nltk.download('punkt')
nltk.download('stopwords')

# Load dataset
df = pd.read_csv("twitter_training.csv")  # Ensure this file is in the working directory

# Check available column names
print("Available columns:", df.columns)  # Print actual column names to verify

# Identify correct column name for text processing
column_name = df.columns[-1]  # Assuming the last column contains text messages

# Initialize Stemmer
stemmer = PorterStemmer()

# Clean text: Remove punctuation & extra spaces
cleaned = []
for text in df[column_name]:  
    cleaned_text = re.sub(r'[^\w\s]', '', str(text))  # Remove punctuation
    cleaned_text = re.sub(r'\s+', ' ', cleaned_text).strip()  # Remove extra spaces
    cleaned.append(cleaned_text)

# Tokenization
tokens = [word_tokenize(x) for x in cleaned]

# Stopword Removal
stop = set(stopwords.words('english'))
filtered_tokens = []
for token_list in tokens:
    filtered_tokens.append([word for word in token_list if word.lower() not in stop])

# Apply Stemming
stemmed_tokens = []
for token_list in filtered_tokens:
    stemmed_tokens.append([stemmer.stem(word) for word in token_list])

# Convert lists of words back into full sentences
stemmed_sentences = [" ".join(tokens) for tokens in stemmed_tokens]

# Initialize CountVectorizer
cv = CountVectorizer(max_features=2000)

# Fit CountVectorizer on preprocessed text data
cv.fit(stemmed_sentences)

# Print the vocabulary
print("Top words in vocabulary:", list(cv.vocabulary_.keys())[:50])  # Print first 50 words
```

```python
import pickle 
with open('vectorizer.pickle','wb') as cvfile:
    pickle.dump(cv, cvfile)
```

```python
x=cv.transform(stemmed_sentences).toarray()
```
